```python
import pandas as pd
from pathlib import Path
```

```python
def txt_to_excel(txt_path, xlsx_path, column_name='text', encoding='utf-8', errors='replace'):
    txt_path = Path(txt_path)
    xlsx_path = Path(xlsx_path)
    if not txt_path.exists():
        raise FileNotFoundError(f"Text file not found: {txt_path}")
    with txt_path.open('r', encoding=encoding, errors=errors) as f:
        lines = [line.rstrip('\n') for line in f]
    df = pd.DataFrame(lines, columns=[column_name])
    df.to_excel(xlsx_path, index=False)
    return xlsx_path
```

```python
# Example: set paths and run
txt_path = input('Path to txt file (or press Enter for example.txt): ').strip() or 'example.txt'
xlsx_path = input('Output xlsx path (or press Enter for output.xlsx): ').strip() or 'output.xlsx'
try:
    out = txt_to_excel(txt_path, xlsx_path)
    print('Saved:', out)
except Exception as e:
    print('Error:', e)
```
